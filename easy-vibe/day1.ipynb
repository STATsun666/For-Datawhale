# 第一章编程入门

### 如何学习本课程

产品工程师这个概念很吸引我。我认为这是把商科和一定的技术能力去交叉在一起，因为实现正在变得简单，token is cheap但是idea怎么来，想法很重要，怎么表述idea也很重要。

我是通过第一章才了解到现在的三种职业：Product Engineer、FDE 和 OPC，我研究生学的是金融科技，为了有差异化优势或者更贴合金融场景，感觉产品岗也很有意思。发现问题和需求，知悉痛点，做市场调查分析，根据结果去做产品。FDE给我的感觉有点像现在四大行里外派出去的审计。

我认为能维护的代码真的是个非常重要的点，哪里是读训练数据和测试数据，哪里是设计损失函数，哪里是读底座模型，哪里调超参数，都很清晰，好调试才是好的代码，像连远程的服务器，对代码的这种维护要求更高。所以我要学如何引导 AI 生成高质量代码

### AI 时代的编程初体验

课件里有一个地方写的很好，他很准确的说出了为什么有的时候ai他做的这个东西就是差点意思，小的规模和清晰的指令很重要。面对大型项目要更关注一个整体上的流程，我的理解是像运筹学一样可以像多阶段问题，什么then什么。

提示词工程也是个很重要的学问，我一般会在自然语言讲清楚我的要求后，去workbuddy自带的增强提示词进行增强，然后再去修改。前端是用户能看到、点到的所有内容；后端是运行在服务器上的数据处理
