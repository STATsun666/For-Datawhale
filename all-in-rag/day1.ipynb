### 学习心得

RAG检索增强生成技术，使AI回答问题之前，先触发检索机制，去外部知识库里把相关的资料段落找出来，当作参考资料给大模型；大模型再结合这些资料和自己的理解生成答案，从而让回答既准确又有时效性。

RAG通过引入外部知识库，丰富上下文信息，区别于微调直接修改模型参数。RAG能解决LLM的痛点，如静态知识局限、幻觉、领域专业性不足、数据隐私风险。（1）准确性与可信度的双重提升（2）时效性保障（3）显著的综合成本效益（4）灵活的模块化可扩展性


基于LangChain框架的RAG实现中：1.使用TextLoader加载外部文件（以路径方式）作为知识源；2.默认分割文本块，保留文本的语义结构；3.将分割后的文本块通过embedding嵌入模型构建向量索引；4.用户交互后发生查询，查看向量相似度，反馈文本块供LLM选取

### 低代码（基于 LlamaIndex）代码解析与注释

在 RAG 流程中，LlamaIndex 将“文档加载 -> 文本切分 -> 向量化与索引构建 -> 检索与生成”进行了高度封装。以下是添加了详细注释的实现代码：

```python
import os
# 配置 Hugging Face 国内镜像源，解决国内网络无法直接下载模型的问题
# os.environ['HF_ENDPOINT']='[https://hf-mirror.com](https://hf-mirror.com)'

from dotenv import load_dotenv
# 从 llama_index 核心库导入：向量存储索引、简易目录文档读取器、全局配置对象
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
# 导入兼容 OpenAI API 格式的第三方大模型接口封装类
from llama_index.llms.openai_like import OpenAILike
# 导入 HuggingFace 本地嵌入（Embedding）模型封装类
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

# 加载项目根目录或当前目录下的 .env 环境变量文件（用于读取 API_KEY 等敏感信息）
load_dotenv()

# 全局配置大语言模型（LLM），用于后续根据检索内容生成自然语言回答
Settings.llm = OpenAILike(
    model="glm-4.7-flash-free",              # 指定调用的模型名称
    api_key=os.getenv("DEEPSEEK_API_KEY"),   # 从环境变量中获取 API 密钥
    api_base="[https://aihubmix.com/v1](https://aihubmix.com/v1)",      # 指定第三方 API 聚合平台的基础接口地址
    is_chat_model=True                       # 声明该模型为对话（Chat）模式模型
)

# 全局配置嵌入模型（Embedding Model），使用 BAAI 开源的轻量级中文向量模型将文本转化为向量
Settings.embed_model = HuggingFaceEmbedding("BAAI/bge-small-zh-v1.5")

# 使用 SimpleDirectoryReader 按指定文件路径加载本地 Markdown 文档，返回 Document 对象列表
documents = SimpleDirectoryReader(
    input_files=["../../data/C1/markdown/easy-rl-chapter1.md"]
).load_data()

# 从加载的文档构建向量索引：
# 内部自动完成文档分块（Node 解析）、调用 Settings.embed_model 计算向量，并存入内存向量库
index = VectorStoreIndex.from_documents(documents)

# 将向量索引转换为查询引擎（Query Engine），默认封装了“相似度检索 + 提示词拼接 + LLM 生成”完整流程
query_engine = index.as_query_engine()

# 打印查询引擎内部默认使用的提示词模板（Prompt Templates），查看其上下文拼接逻辑
print(query_engine.get_prompts())

# 执行 RAG 查询：先检索与问题最相关的文档片段，再交由 Settings.llm 生成最终回答并打印
print(query_engine.query("文中举了哪些例子?"))
```
